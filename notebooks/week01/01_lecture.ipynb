## S2 · 列表、元组与字典

### 列表（list）：有顺序的一串值

**Python 从 0 开始数**：第一个元素的位置（索引）是 0。

```python
shopping = ["apple", "milk", "bread", "eggs"]
print(shopping[0])     # 第一个
print(shopping[-1])    # 最后一个
print(len(shopping))   # 有几个
shopping.append("tea") # 在末尾加一个
print(shopping)
```

**切片** `[开始:结束]` 取出一段，注意“包含开始、不包含结束”：

```python
print(shopping[1:3])
print(shopping[:2])
```

## S3 · 循环与条件判断

### for 循环：对每一个元素做同一件事

```python
for item in shopping:
    print(item)
```

Python 用**缩进**（4个空格）表示哪些代码属于循环。`enumerate` 同时给出位置和值，`zip` 把两个列表配对：

```python
for i, item in enumerate(shopping):
    print(i, item)

for name, row in zip(names[:5], scores[:5]):
    print(name, row)
```
